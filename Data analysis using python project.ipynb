```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
categorical_cols = student.select_dtypes(
    include="object"
).columns
```

```python
# Loop through each categorical column
for col in categorical_cols:
    # Create count plot
    sns.countplot(x=col, data=student)
    # Rotate x-axis labels
    plt.xticks(rotation=45)
    # Set plot title
    plt.title(f"Count of {col}")
    # Adjust layout
    plt.tight_layout()
    # Display plot
    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Loop through all columns
for col in student:
    # Check whether the column is numerical
    if student[col].dtype != "object":
        # Plot histogram with KDE
        sns.histplot(student[col], kde=True)
        # Set title and labels
        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        # Adjust layout
        plt.tight_layout()
        # Display plot
        plt.show()
```

```python
# Hours studied vs exam score
plt.figure(figsize=(8, 8))
sns.lmplot(
    x="Hours_Studied",
    y="Exam_Score",
    data=student
)
plt.show()
```

```python
# Parental involvement vs exam score
sns.boxplot(
    x="Parental_Involvement",
    y="Exam_Score",
    data=student
)
plt.title("Parental Involvement vs Exam Score")
plt.show()
```

```python
# Family income vs exam score
sns.violinplot(
    x="Family_Income",
    y="Exam_Score",
    data=student
)
# Violin plot combines:
# 1. Box plot (median, quartiles and outliers)
# 2. Kernel density plot (distribution shape)
plt.title("Family Income vs Exam Score")
plt.show()
```

```python
# Select columns for pair plot
selected_columns = [
    "Hours_Studied",
    "Attendance",
    "Previous_Scores",
    "Sleep_Hours"
]
# Create pairwise scatter plots and histograms
sns.pairplot(student[selected_columns])
# Add main title
plt.suptitle(
    "Pair Plot of Selected Variables",
    y=1.02
)
# Display plot
plt.show()
```

```python
# Gender vs exam score
sns.barplot(
    x="Gender",
    y="Exam_Score",
    data=student
)
plt.title("Gender vs Exam Score")
plt.show()
```

```python
sns.lmplot(
    x="Sleep_Hours",
    y="Exam_Score",
    data=student
)
plt.title("Sleep Hours vs Exam Score")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6, 4))
# Boxplot comparing exam scores by gender
sns.boxplot(
    x="Gender",
    y="Exam_Score",
    data=student,
    hue="Gender",
    palette="Set2"
)
# Add titles and labels
plt.title("Exam Score Comparison: Boys vs Girls")
plt.xlabel("Gender")
plt.ylabel("Exam Score")
# Prevent clipping
plt.tight_layout()
# Display plot
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Select numerical features
numerical_cols = [
    "Hours_Studied",
    "Attendance",
    "Sleep_Hours",
    "Previous_Scores",
    "Tutoring_Sessions",
    "Physical_Activity",
    "Exam_Score"
]
# Calculate correlation
corr = student[numerical_cols].corr()
# Create heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Create 2x2 subplot grid
fig, axes = plt.subplots(
    2, 2,
    figsize=(12, 8)
)
# Plot 1: Access to Resources
sns.stripplot(
    x="Access_to_Resources",
    y="Exam_Score",
    data=student,
    palette="Set2",
    ax=axes[0, 0]
)
axes[0, 0].set_title(
    "Exam Score vs Access to Resources"
)
axes[0, 0].set_xlabel("Access to Resources")
axes[0, 0].set_ylabel("Exam Score")
# Plot 2: Extracurricular Activities
sns.stripplot(
    x="Extracurricular_Activities",
    y="Exam_Score",
    data=student,
    palette="Set3",
    ax=axes[0, 1]
)
axes[0, 1].set_title(
    "Exam Score vs Extracurricular Activities"
)
axes[0, 1].set_xlabel("Extracurricular Activities")
axes[0, 1].set_ylabel("Exam Score")
# Plot 3: Motivation Level
sns.stripplot(
    x="Motivation_Level",
    y="Exam_Score",
    data=student,
    palette="husl",
    ax=axes[1, 0]
)
axes[1, 0].set_title(
    "Exam Score vs Motivation Level"
)
axes[1, 0].set_xlabel("Motivation Level")
axes[1, 0].set_ylabel("Exam Score")
# Plot 4: Teacher Quality
sns.stripplot(
    x="Teacher_Quality",
    y="Exam_Score",
    data=student,
    palette="hls",
    ax=axes[1, 1]
)
axes[1, 1].set_title(
    "Exam Score vs Teacher Quality"
)
axes[1, 1].set_xlabel("Teacher Quality")
axes[1, 1].set_ylabel("Exam Score")
# Adjust layout
plt.tight_layout()
# Display all plots
plt.show()
```
